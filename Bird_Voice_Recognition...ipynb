```python
import pandas as pd

csv_path = "Extracted_Dataset/Birds Voice.csv"

df = pd.read_csv(csv_path)

print(df.head())
print(df.columns)
print(df.shape)
```

```python
df["common_name"].value_counts().head(20)
```

```python
print("Total bird classes:", df["common_name"].nunique())
```

```python
#bird name
bird_names = sorted(df["common_name"].dropna().unique())

for i, bird in enumerate(bird_names, start=1):
    print(i, bird)
```

```python
import matplotlib.pyplot as plt

df["common_name"].value_counts().plot(
    kind="bar",
    figsize=(18, 6)
)

plt.title("Number of Recordings per Bird Species")
plt.xlabel("Bird Species")
plt.ylabel("Number of Recordings")
plt.xticks(rotation=90)
plt.show()
```

```python
selected_birds = [
    "Andean Guan",
    "Andean Tinamou",
    "Australian Brushturkey",
    "Band-tailed Guan"
]

bird_df = df[
    df["common_name"].isin(selected_birds)
].copy()

print("Selected recordings:", len(bird_df))
print(bird_df["common_name"].value_counts())
```

```python
bird_df.to_csv(
    "selected_4_birds.csv",
    index=False
)

print("Selected 4-bird CSV saved successfully!")
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

bird_df["label"] = label_encoder.fit_transform(
    bird_df["common_name"]
)

print(bird_df[["common_name", "label"]].head())
print(label_encoder.classes_)
```

```python
label_mapping = {
    index: bird
    for index, bird in enumerate(label_encoder.classes_)
}

print(label_mapping)
```
