```python
import os
import re
import numpy as np
import pandas as pd
import librosa
import librosa.display
import matplotlib.pyplot as plt

# Paths
PROJECT_FOLDER = r"C:\Users\Nikhil Prajapati\BirdVoiceRecognition"
CSV_PATH = os.path.join(PROJECT_FOLDER, "selected_4_birds.csv")

OUTPUT_FOLDER = os.path.join(
    PROJECT_FOLDER,
    "Spectrogram_Dataset"
)

# Load CSV
df = pd.read_csv(CSV_PATH)

# Create output folder
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

print("Output folder:", OUTPUT_FOLDER)
```

```python
def create_mel_spectrogram(audio_path, output_path):
    try:
        # Load audio
        y, sr = librosa.load(audio_path, sr=22050)

        # Create Mel Spectrogram
        mel = librosa.feature.melspectrogram(
            y=y,
            sr=sr,
            n_mels=128
        )

        # Convert to decibel
        mel_db = librosa.power_to_db(
            mel,
            ref=np.max
        )

        # Create image
        plt.figure(figsize=(4, 4))

        librosa.display.specshow(
            mel_db,
            sr=sr,
            x_axis=None,
            y_axis=None
        )

        plt.axis("off")
        plt.tight_layout(pad=0)

        # Save image
        plt.savefig(
            output_path,
            bbox_inches="tight",
            pad_inches=0
        )

        plt.close()

        return True

    except Exception as e:
        print("Error:", audio_path, e)
        return False
```

```python
processed_count = 0
failed_count = 0

for audio_path in audio_files:

    file_name = os.path.splitext(
        os.path.basename(audio_path)
    )[0]

    class_folder = os.path.join(
        OUTPUT_FOLDER,
        "all_audio"
    )

    os.makedirs(class_folder, exist_ok=True)

    output_path = os.path.join(
        class_folder,
        file_name + ".png"
    )

    if os.path.exists(output_path):
        processed_count += 1
        continue

    success = create_mel_spectrogram(
        audio_path,
        output_path
    )

    if success:
        processed_count += 1
    else:
        failed_count += 1

print("Spectrogram generation completed!")
print("Processed:", processed_count)
print("Failed:", failed_count)
```
